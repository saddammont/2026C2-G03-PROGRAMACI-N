```python
# Escriba su solución aquí.
registros = df.to_dict(orient="records")
print(registros[0:5])#Muestra el primer registro como ejemplo
# registros = ...
contador = 0 #almacena la cantidad de registros superiores a 3000

for registro in registros:
    if registro["llegadas_miles"] > 3000:
        contador += 1

print(f"Cantidad de registros con llegadas_miles superiores a 3000: {contador}")

# contador = ...

def resumen_pais(registros, pais):
    # Recorra los registros y acumule solo los del país indicado.
    visitas = 0
    for lugar in registros:
        if lugar["pais"] == pais:
            visitas += lugar["visitantes"]
    return visitas

# Pruebe la función con 'Costa Rica' y con un país inexistente.
print(resumen_pais(registros, "Costa Rica"))
print(resumen_pais(registros, "pais no exiaste"))  # país inexistente 
```

```python
df_2019 = df[df["anio"] == 2019].set_index("pais")["llegadas_miles"]
df_2020 = df[df["anio"] == 2020].set_index("pais")["llegadas_miles"]

#display(df_2019)


df_analisis = pd.DataFrame({"llegadas_miles_2019": df_2019, "llegadas_miles_2020": df_2020}).dropna()

df_analisis["caida porcentual"] = (df_analisis["llegadas_miles_2019"] - df_analisis["llegadas_miles_2020"]) / df_analisis["llegadas_miles_2019"] * 100


display(df_analisis.sort_values("caida porcentual", ascending=False)
        .sort_values("caida porcentual", ascending=False).head(10)
        .style.format({"caida porcentual": "{:.2f}%"}
        ))
```
